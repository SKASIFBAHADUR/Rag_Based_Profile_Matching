## Simple Evaluation
Manually define expected relevant candidates for each JD and calculate Precision@10 / Recall@10.

```python
def precision_recall_at_k(retrieved, relevant, k=10):
    retrieved_k = retrieved[:k]
    hits = len(set(retrieved_k) & set(relevant))
    precision = hits / k if k else 0
    recall = hits / len(relevant) if relevant else 0
    return precision, recall

retrieved = [m['candidate_name'] for m in result['top_matches']]
relevant = ['Candidate 01', 'Candidate 07', 'Candidate 13', 'Candidate 19', 'Candidate 25']
precision_recall_at_k(retrieved, relevant, k=10)
```
